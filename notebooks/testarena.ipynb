```python
test_random_bot_only_plays_legal_moves()
```

```python
def test_bots_do_not_mutate_the_board():
    board = chess.Board()
    fen_before = board.fen()
    MaterialBot(seed=0).select_move(board)
    assert board.fen() == fen_before


```

```python
test_bots_do_not_mutate_the_board()
```

```python
def test_material_bot_finds_mate_in_one():
    board = chess.Board("6k1/5ppp/8/8/8/8/8/R5K1 w - - 0 1")
    move = MaterialBot(seed=0).select_move(board)
    board.push(move)
    assert board.is_checkmate()
```

```python
test_material_bot_finds_mate_in_one()
```
